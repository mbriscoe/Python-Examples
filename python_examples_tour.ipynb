## 2. Core data structures

### Lists

<details>
<summary>Explanation</summary>

Lists are ordered and mutable. This example creates, updates, removes, searches, measures, and iterates over list values.

</details>

```python
my_list = []
my_list.extend([1, 2, 3])
print(my_list[0], my_list[1])
my_list[2] = 4
print(my_list)
my_list.remove(2)
print(my_list)
print(1 in my_list)
print(len(my_list))
for item in my_list:
    print(item)
```

### Dictionaries

<details>
<summary>Explanation</summary>

Dictionaries map keys to values. Use square brackets when a missing key should be an error, or `.get` when a default or `None` is preferable.

</details>

```python
student = {
    "name": "John",
    "age": 20,
    "major": "Computer Science"
}
print(student["name"])
print(student.get("age"))
student["age"] = 21
student["university"] = "ABC University"
del student["major"]
for key, value in student.items():
    print(key, ":", value)
print("name" in student)
print(len(student))
```

### Sets and tuples

<details>
<summary>Explanation</summary>

Sets hold unique values and support mathematical operations. Tuples are ordered and immutable, which makes them useful for fixed groups of values such as a card's rank and suit.

</details>

```python
my_set = {1, 2, 3, 4, 5, 2}
my_set.add(6)
my_set.remove(3)
print("Set:", my_set)
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference:", set1 - set2)
print("Symmetric difference:", set1 ^ set2)

my_tuple = (1, 2, 3, "a", "b", "c")
print(my_tuple[0], my_tuple[3])
print(my_tuple[2:5])
a, b, c, d, e, f = my_tuple
print(a, b, c, d, e, f)
print(len(my_tuple), "a" in my_tuple)
```

## 3. Functions and expressive syntax

### `*args`, `**kwargs`, lambdas, and comprehensions

<details>
<summary>Explanation</summary>

`*args` collects positional arguments into a tuple, while `**kwargs` collects named arguments into a dictionary. A lambda is a small anonymous function. A list comprehension expresses a loop and filter in one readable expression.

</details>

```python
def sum_numbers(*args):
    total = 0
    for number in args:
        total += number
    return total

def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print(sum_numbers(1, 2, 3))
print(sum_numbers(10, 20, 30, 40))
print_info(name="John", age=25, city="New York")

multiply = lambda x, y: x * y
print(multiply(2, 3))

number_list = [12, 45, 60, 87, 999, 200, 85, 77, 2, 3]
odd_numbers = [number for number in number_list if number % 2]
print(odd_numbers)
```

### Decorators

<details>
<summary>Explanation</summary>

A decorator wraps a function to add behavior without changing the function's main implementation. The `@add_author` syntax applies the wrapper when the function is defined.

</details>

```python
def add_author(function):
    def wrapper(*args):
        function_result = function(*args)
        return f"{function_result}\nBy Mark Briscoe"
    return wrapper

@add_author
def quote_of_the_day(quote):
    return "Quote of the day: " + quote

print(quote_of_the_day("We love mixins!"))
```

## 4. Classes, inheritance, and algorithms

### Classes and inheritance

<details>
<summary>Explanation</summary>

`Bird` defines shared state and behavior. `BigBird` and `ReallyBigBird` inherit from it, call the parent constructor with `super`, and customize `describe`.

</details>

```python
class Bird:
    number_of_birds = 0

    def __init__(self, kind, call):
        self.kind = kind
        self.call = call
        Bird.number_of_birds += 1

    def describe(self):
        return f"{self.kind} goes {self.call}."

class BigBird(Bird):
    def __init__(self, kind, call, wingspan):
        self.wingspan = wingspan
        super().__init__(kind, call)

    def describe(self):
        return f"{self.kind} goes {self.call}, and has a wingspan of {self.wingspan} feet!"

class ReallyBigBird(BigBird):
    def __init__(self, kind, call, wingspan, age):
        self.age = age
        super().__init__(kind, call, wingspan)

    def describe(self):
        return f"{self.kind} goes {self.call}, and has a wingspan of {self.wingspan} feet! It is {self.age} years old and massive!"

vulture = ReallyBigBird("Vulture", "Squawk", 60, 200)
print(vulture.describe())
print("Birds created:", Bird.number_of_birds)
```

### FizzBuzz

<details>
<summary>Explanation</summary>

Check divisibility by 15 before checking 3 or 5. The compact version builds the result from independent conditions and falls back to the number.

</details>

```python
def fizz_buzz_short(number):
    return f"{'' if number % 3 else 'Fizz'}{'' if number % 5 else 'Buzz'}" or number

for number in range(1, 16):
    print(fizz_buzz_short(number))
```

### Palindromes and prime numbers

<details>
<summary>Explanation</summary>

A palindrome reads the same forwards and backwards; slicing with `[::-1]` creates the reversed string. To test a prime, try divisors only up to the square root of the number.

</details>

```python
def is_palindrome(text):
    return text == text[::-1]

for word in ["racecar", "python"]:
    print(word, is_palindrome(word))

primes_below_30 = []
for number in range(2, 30):
    if all(number % divisor for divisor in range(2, int(number ** 0.5) + 1)):
        primes_below_30.append(number)
print(primes_below_30)
```

### Bubble sort

<details>
<summary>Explanation</summary>

Bubble sort repeatedly compares neighboring values and swaps them when they are out of order. It sorts the list in place.

</details>

```python
def bubble_sort(numbers):
    for pass_number in range(len(numbers) - 1, 0, -1):
        for index in range(pass_number):
            if numbers[index] > numbers[index + 1]:
                numbers[index], numbers[index + 1] = numbers[index + 1], numbers[index]

numbers = [14, 46, 43, 27, 57, 41, 45, 21, 70]
bubble_sort(numbers)
print(numbers)
```

### Card-game simulation

<details>
<summary>Explanation</summary>

The original card game waits for input between hands and uses console colors. This notebook version keeps the same deck, deal, comparison, and scoring ideas but simulates a chosen number of hands without prompts.

</details>

```python
import random

suits = ("Hearts", "Clubs", "Diamonds", "Spades")
numbers = range(2, 15)

def create_deck():
    deck = [(number, suit) for suit in suits for number in numbers]
    random.shuffle(deck)
    return deck

def process_card(card_number):
    names = {11: "Jack", 12: "Queen", 13: "King", 14: "Ace"}
    return names.get(card_number, card_number)

random.seed(7)
deck = create_deck()
player1_pack = []
player2_pack = []

for hand_number in range(5):
    player1_card = deck.pop()
    player2_card = deck.pop()
    print(f"Hand {hand_number + 1}: {process_card(player1_card[0])} of {player1_card[1]} vs "
          f"{process_card(player2_card[0])} of {player2_card[1]}")
    if player1_card[0] > player2_card[0]:
        player1_pack.extend([player1_card, player2_card])
        print("Player 1 wins")
    elif player1_card[0] < player2_card[0]:
        player2_pack.extend([player1_card, player2_card])
        print("Player 2 wins")
    else:
        player1_pack.append(player1_card)
        player2_pack.append(player2_card)
        print("Draw")

print(f"Scores: Player 1 = {len(player1_pack)}, Player 2 = {len(player2_pack)}")
```
